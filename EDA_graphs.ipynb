```python
#  UNIVARIATE GRAPHS

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns



# Histogram for Sales
plt.figure(figsize=(8, 6))  # Adjust figure size for better visualization
sns.histplot(data=df, x="Sales", kde=True) # Added KDE for smoother distribution representation
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()

# Histogram for Profit
plt.figure(figsize=(8, 6))
sns.histplot(data=df, x="Profit", kde=True)
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()

# Line Plot for Sales by Category
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Category", y="Sales")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

# Bar Plot for Sales by Sub-Category
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x="Sub-Category", y="Sales")
plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for readability
plt.show()

# Bar Plot for Profit by Category (improved)
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Category', y='Profit')
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()

# Boxplot for Sales by Category
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Category', y='Sales')
plt.title("Sales Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

# Boxplot for Profit by Category
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Category',y='Profit')
plt.title("Profit Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
#  BIVARIATE GRAPHS

# Scatter Plot for Sales vs. Profit
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="Sales", y="Profit")
plt.title("Sales vs. Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()

# Joint Plot for Sales vs. Profit (with KDE)
plt.figure(figsize=(8, 6))
sns.pairplot(data=df, x="Sales", y="Profit", kind="kde") # Use kind="kde" for density plot
plt.suptitle("Sales vs. Profit (KDE)", y=1.02) # Adjust title position
plt.show()

# Pair Plot for selected numerical columns
numerical_cols = ['Sales', 'Quantity', 'Discount', 'Profit']
sns.pairplot(df[numerical_cols])
plt.suptitle("Pairwise Relationships of Numerical Variables", y=1.02)
plt.show()

# Heatmap for correlation matrix
plt.figure(figsize=(10,8))
correlation_matrix = df[numerical_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix of Numerical Variables")
plt.show()

# Bubble Plot (Sales vs Profit, with Quantity as bubble size)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Sales', y='Profit', size='Quantity', sizes=(20, 200), hue='Quantity')
plt.title('Sales vs. Profit (Bubble Plot)')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```
